```python
df['text'] = df['text'].apply(lambda x : x.lower())
```

```python
import string
def remove_punc(txt):
    return txt.translate(str.maketrans('','',string.punctuation))
```

```python
df['text'] = df['text'].apply(remove_punc)
```

```python
import re

df["text"] = df["text"].apply(lambda x: re.sub(r"\d+", "", x))
```

```python
df
```

```python
def remove_emoji(txt):
    new = ""
    for ch in txt:
        if ch.isascii():
            new += ch
    return new

df["text"] = df["text"].apply(remove_emoji)
```

```python
from nltk.tokenize import word_tokenize

def remove(txt):
    words = word_tokenize(txt)
    cleaned = []

    for word in words:
        if word not in stop_words:
            cleaned.append(word)

    return " ".join(cleaned)

df["text"] = df["text"].apply(remove)
```

# 🔥 Best Model Selection (Final)

This section trains multiple NLP classifiers on the TF-IDF features and automatically selects the highest accuracy model.

**Models compared**
- Linear SVM ⭐
- Logistic Regression
- SGD Classifier
- Multinomial Naive Bayes
- Random Forest

The notebook prints the accuracy of each model, displays a sorted comparison table, and keeps the best model for prediction.


```python
from sklearn.svm import LinearSVC
from sklearn.linear_model import LogisticRegression, SGDClassifier
from sklearn.naive_bayes import MultinomialNB
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score, classification_report
import pandas as pd
import joblib

# Optimized TF-IDF
from sklearn.feature_extraction.text import TfidfVectorizer

tfidf_vectorizer = TfidfVectorizer(
    lowercase=True,
    stop_words='english',
    ngram_range=(1,2),
    min_df=2,
    max_df=0.95,
    sublinear_tf=True
)

X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)
X_test_tfidf = tfidf_vectorizer.transform(X_test)

models = {
    "Linear SVM": LinearSVC(),
    "Logistic Regression": LogisticRegression(max_iter=1000),
    "SGD Classifier": SGDClassifier(random_state=42),
    "Multinomial NB": MultinomialNB(),
    "Random Forest": RandomForestClassifier(
        n_estimators=200,
        random_state=42
    )
}

results = {}
trained_models = {}

for name, model in models.items():
    model.fit(X_train_tfidf, y_train)
    pred = model.predict(X_test_tfidf)
    acc = accuracy_score(y_test, pred)

    results[name] = acc
    trained_models[name] = model

    print(f"{name:22} : {acc:.4f}")

result_df = pd.DataFrame({
    "Model": results.keys(),
    "Accuracy": results.values()
}).sort_values(by="Accuracy", ascending=False)

print("\n🏆 Best Model")
best_model_name = result_df.iloc[0]["Model"]
best_model = trained_models[best_model_name]

print("Model :", best_model_name)
print("Accuracy :", result_df.iloc[0]["Accuracy"])

display(result_df)

# Detailed report
best_pred = best_model.predict(X_test_tfidf)
print("\nClassification Report\n")
print(classification_report(y_test, best_pred))

# Save model + vectorizer
joblib.dump(best_model, "best_nlp_model.pkl")
joblib.dump(tfidf_vectorizer, "tfidf_vectorizer.pkl")

print("\n✅ Saved: best_nlp_model.pkl")
print("✅ Saved: tfidf_vectorizer.pkl")
```
